### 참고
- `int64` : 정수. `RESPOND_ID`, 일자 컬럼들(`EXAMIN_BEGIN_DE_x/y`, `TOUR_BEGIN_DE`)은 값이 전부 숫자로만 되어있어 pandas가 자동으로 정수로 인식함
- `object` : 문자열(또는 혼합 타입). 지역명·기간값·지출금액 등 대부분이 여기 해당함
- `float64` : 실수. `COM_SIX_TY`(동행자6유형)는 값이 전부 결측이라 pandas가 임의로 float64로 인식함 (실제 의미상 문자열이어야 하지만 값이 하나도 없어 그렇게 판단된 것)
- 공식 컬럼정의서(문화빅데이터 플랫폼)상으로는 38개 전부 `VARCHAR`(문자열)로 설계됨 → pandas가 int64/float64로 자동 변환한 건 CSV 특성상 값을 보고 추론한 결과이며, 원본 설계와는 다를 수 있음
- 지출 관련 컬럼(여행총비용값 등)이 `object`로 남아있는 이유는 아직 미확인 → 실제 값 확인 예정
- _x, _y 쌍으로 존재하는 응답자 특성 컬럼(성별·연령대·직업 등)은 동일 응답자가 두 조사에서 같은 값으로 응답했는지 추후 코드로 일치 여부 검증 필요함

## [참고] 결측치를 찾는 방법 
<div style="font-size: 12px; line-height: 1.3;">

### 결측치 유형 6가지

| 유형 | 설명 | 발견 방법 | 이번 프로젝트 예시 |
|---|---|---|---|
| A. 눈에 보이는 빈 칸 | pandas가 `NaN`으로 이미 표시해준 것 | `df.info()`, `df.isnull().sum()` | `COM_ONE_TY`~`COM_SIX_TY` |
| B. 글자로 위장한 결측치 | `Non-Null`로 잡히지만 실제 의미는 결측 (`'모름'`, `'무응답'`, `'-'` 등) | `.unique()`로 실제 값 열어보기 | `TOUR_TOT_CT_VALUE`의 `'모름'` |
| C. 플레이스홀더(placeholder) 숫자 | 말도 안 되는 숫자로 결측 표시 (`999`, `-1`, `00000000` 등) | `.describe()`의 min/max, `.unique()`로 이상한 값 확인 | (이번 데이터엔 미발견) |
| D. 형변환 시 드러나는 결측치 | 숫자로 강제 변환(`pd.to_numeric` 등) 시도할 때 변환 안 되는 값들이 `NaN`으로 쏟아짐 | `pd.to_numeric(df[col], errors='coerce')` 후 `.isnull().sum()` 비교 | 지출 컬럼 숫자 변환 시 재확인 필요 |
| E. 다른 컬럼과 비교해야 보이는 논리적 결측치 | 값은 있지만 다른 컬럼과 모순되어 신뢰 불가 | `groupby`, `crosstab`으로 컬럼 간 교차 확인 | 아직 미확인, 필요 시 확인 |
| F. 시각화로 패턴 파악 | 결측치가 특정 컬럼·특정 행 구간에 몰려있는지 그림으로 확인 | `missingno` 라이브러리(`msno.matrix(df)`) | 컬럼이 아주 많을 때 특히 유용 |

### 결측치를 발견했을 때 스스로 던지는 4가지 질문

1. **"이 빈 칸이 왜 생겼을까?"** — 구조적으로 당연히 없는 값인지, 원래 있어야 하는데 누락된 값인지 구분
2. **"이 빈 칸을 그냥 두면 나중에 무슨 문제가 생길까?"** — 실제로 이 컬럼을 어디에 쓸지 생각해보고 결측이 방해되는지 확인
3. **"뭘로 채우거나 어떻게 처리해야 데이터의 진짜 의미가 안 왜곡될까?"** — 구조적 결측은 의미 있는 값으로 채우고, 진짜 누락은 임의로 채우지 않음
4. **"처리 후에도 데이터가 말이 되는가?"** — 처리 후 다시 확인해서 비정상적인 값이 새로 생기지 않았는지 검증


```python
df['TOUR_TOT_CT_VALUE'].unique()[:20]
#지출 컬럼이 왜 문자열인지 확인

#출력값에서 '모름'이라는 명시적인 결측/무응답 값 섞여있음 
```

```python
#_x/_y 쌍 컬럼들의 일치율 먼저 확인
x_cols = [col for col in df.columns if col.endswith('_x')]

for x_col in x_cols:
    base = x_col[:-2]           # '_x' 떼고 원래 이름 얻기
    y_col = base + '_y'
    if y_col in df.columns:
        match_rate = (df[x_col].fillna('__NA__') == df[y_col].fillna('__NA__')).mean() #.mean(): True(1)/False(0)의 평균
        print(f'{base} : 일치율 {match_rate*100:.1f}%') #일치하는 비율 구하기 

        # NaN == NaN은 파이썬에서 True가 아니라 False로 나옴
        # 결측치끼리는 "같다"고 인식이 안 됨, 이걸 방지하려고 둘 다 같은 임시 문자열('__NA__')로 잠깐 바꿔서 비교
```

```python
#100% 일치하는 쌍만 병합하기
full_duplicate = []

for x_col in x_cols:
    base = x_col[:-2]
    y_col = base + '_y'
    if y_col in df.columns:
        match_rate = (df[x_col].fillna('__NA__') == df[y_col].fillna('__NA__')).mean()
        if match_rate == 1.0:
            full_duplicate.append(base)

print('완전 일치해서 병합할 컬럼:', full_duplicate)
print('병합 전 컬럼 수:', df.shape[1])

drop_cols = [name + '_y' for name in full_duplicate]
df = df.drop(columns=drop_cols)
df = df.rename(columns={name + '_x': name for name in full_duplicate})

print('병합 후 컬럼 수:', df.shape[1])
```

```python
#'모름'이라는 값을 가진 컬럼 구별하기 위한 코드 

suspect_word = '모름'

found = []
clean = []

for col in df.columns:
    # dtype 상관없이 일단 전부 문자열로 변환해서 검사 (숫자형이어도 확인함)
    vals = df[col].dropna().astype(str).unique()

    if any(suspect_word in v for v in vals):
        cnt = df[col].astype(str).value_counts().get(suspect_word, 0)
        found.append((col, cnt))
    else:
        clean.append(col)

print(f"'{suspect_word}' 포함된 컬럼: {len(found)}개")
for c, cnt in found:
    print(f"  {c} : {cnt}건 ({cnt/len(df)*100:.1f}%)")

print()
print(f"'{suspect_word}' 없는 컬럼: {len(clean)}개")
for c in clean:
    print(' ', c)
```

- 결측 원인: 구간형 텍스트 응답 + `'모름'`(진짜 무응답)
- 처리: 구간 텍스트 → 대표값(숫자)으로 변환, `'모름'`은 임의로 채우지 않고 결측(NaN)으로 유지

- 처리: 별도 대체 없이 `'모름'`을 하나의 정상 카테고리로 유지 (임의 대체 시 응답자 4명 중 1명 정보 왜곡 우려)

### 행을 지우지말고 데이터 양을 보존함 / '모름'으로 카테고리를 남겨 '모름'이라 답한 사람들에 대해 어떤 특징이 있는지 분석할 수 있음'

```python
group1_cols = ['COM_ONE_TY', 'COM_TWO_TY', 'COM_THREE_TY', 'COM_FOUR_TY', 'COM_FIVE_TY', 'COM_SIX_TY']
for col in group1_cols:
    before = df[col].isnull().sum()
    df[col] = df[col].fillna('해당없음')
    after = df[col].isnull().sum()
    print(f'{col} : 처리 전 결측 {before}건 → 처리 후 {after}건')
```

```python
#그룹 2 변환 코드 
# 총비용 (10만원 단위)
total_map = {
    '10만원 미만': 5, '10~20만원 미만': 15, '20~30만원 미만': 25,
    '30~40만원 미만': 35, '40만원 이상': 45
}

# 세부 항목 6개 (1만원 단위, 전부 동일 체계)
item_map = {
    '1만원 미만': 0.5, '1~3만원 미만': 2, '3~5만원 미만': 4,
    '5~7만원 미만': 6, '7~10만원 미만': 8.5, '10만원 이상': 11.5
}

# 여행기간
pd_map = {
    '1박 2일': 2, '2박 3일': 3, '3박 4일': 4,
    '4박 5일': 5, '5박 6일': 6, '6박 7일 이상': 7
}

# 동행인원 (혼자서=1명으로 매핑)
nmpr_map = {
    '혼자서': 1, '2명': 2, '3명': 3, '4명': 4, '5명 이상': 5
}

item_cols = ['TOUR_LDGMNT_CT_VALUE', 'TOUR_FOOD_CT_VALUE', 'TOUR_TRNSPORT_CT_VALUE',
             'TOUR_SHOPNG_CT_VALUE', 'TOUR_ACTVTY_CT_VALUE', 'TOUR_ETC_CT_VALUE']

df['TOUR_TOT_CT_VALUE_NUM'] = df['TOUR_TOT_CT_VALUE'].map(total_map)
for col in item_cols:
    df[col + '_NUM'] = df[col].map(item_map)
df['TOUR_PD_VALUE_NUM'] = df['TOUR_PD_VALUE'].map(pd_map)
df['TOUR_COM_NMPR_NM_NUM'] = df['TOUR_COM_NMPR_NM'].map(nmpr_map)

check_cols = ['TOUR_TOT_CT_VALUE_NUM'] + [c+'_NUM' for c in item_cols] + ['TOUR_PD_VALUE_NUM', 'TOUR_COM_NMPR_NM_NUM']
print(df[check_cols].isnull().sum())

# 그룹2: 결측 개수가 이전 '모름' 건수와 정확히 일치함을 확인 
```

## 이상치

```python
# 이상치 확인할 컬럼
outlier_check_cols = ['TOUR_TOT_CT_VALUE_NUM', 'TOUR_LDGMNT_CT_VALUE_NUM', 'TOUR_FOOD_CT_VALUE_NUM',
                       'TOUR_TRNSPORT_CT_VALUE_NUM', 'TOUR_SHOPNG_CT_VALUE_NUM', 'TOUR_ACTVTY_CT_VALUE_NUM',
                       'TOUR_ETC_CT_VALUE_NUM', 'TOUR_PD_VALUE_NUM', 'TOUR_COM_NMPR_NM_NUM']

df[outlier_check_cols].describe()
#RESPOND_ID(응답자ID)나 EXAMIN_BEGIN_DE(조사일자) 식별용 숫자라 제외 
```

```python
#df에서 확인하기 어려우니 한 눈에 보기 
for col in outlier_check_cols:
    q1 = df[col].quantile(0.25)
    q3 = df[col].quantile(0.75)
    iqr = q3 - q1
    lower = q1 - 1.5 * iqr
    upper = q3 + 1.5 * iqr
    outlier_count = ((df[col] < lower) | (df[col] > upper)).sum()
    print(f'{col} : Q1={q1}, Q3={q3}, 정상범위=[{lower:.1f}, {upper:.1f}], 이상치 {outlier_count}건')
```

```python
#진짜 극단값이 있는지 확인 
suspect_cols = ['TOUR_SHOPNG_CT_VALUE_NUM', 'TOUR_ACTVTY_CT_VALUE_NUM', 'TOUR_ETC_CT_VALUE_NUM', 'TOUR_PD_VALUE_NUM']

for col in suspect_cols:
    print(col)
    print(df[col].value_counts(dropna=False).sort_index())
    print()
```

- 방법: IQR(사분위범위) 1.5배 기준으로 1차 스캔
- 결과:
  - 총비용·숙박비·음식비·교통비·동행인원: 이상치 0건
  - 쇼핑비·액티비티비·기타비·여행기간: IQR상 이상치 다수 검출되었으나, 값이 매핑표 범위(0.5~11.5 또는 2~7) 내 정상 응답값으로만 구성되어 있음을 `.value_counts()`로 확인함
  - 원인: 응답이 특정 구간(예: 0.5)에 쏠린 분포 특성상 Q1=Q3가 되며 IQR이 비정상적으로 좁아져 발생한 통계적 착시라고 판단
- 처리: 전 컬럼 원본 유지, 삭제/변경 없음

## 손상 레코드 

<div style="font-size: 12px; line-height: 1.3;">

# 확인할 손상 패턴 4가지

1. **시간 순서 모순**
   - 여행일이 조사일보다 늦을 수 없음
   - 확인: `TOUR_BEGIN_DE > EXAMIN_BEGIN_DE`인 경우

2. **동일 응답자 고정 속성 충돌**
   - 같은 `RESPOND_ID`라면 성별, 연령대 같은 값이 바뀌면 안 됨
   - 확인: 동일 응답자인데 `SEXDSTN_FLAG_CD`(성별)가 여러 값인 경우
   - 확인: 동일 응답자인데 `AGRDE_FLAG_NM`(연령대)이 2단계 이상 차이나는 경우

3. **물리적 자원 충돌**
   - 한 사람이 겹치는 기간에 두 여행을 동시에 다녀올 수 없음
   - 확인: 같은 응답자의 서로 다른 두 여행 기간이 겹치는 경우

4. **중복 키**
   - 같은 응답자 + 같은 여행 시작일 조합인데 조사 회차마다 응답이 다르면 안 됨
   - 확인: 동일 응답자+동일 여행시작일이 여러 조사 회차에 걸쳐 응답된 경우
   - 확인: 위 중복 응답 중 여행 지역까지 다르게 기록된 경우

```python
# 시간 순서 모순부터 확인
count_time_violation = (df['TOUR_BEGIN_DE'] > df['EXAMIN_BEGIN_DE']).sum()
print(f'여행일이 조사일보다 늦은 경우: {count_time_violation}건')

# 실제로 몇 건인지, 어떤 모습인지 눈으로 확인
df[df['TOUR_BEGIN_DE'] > df['EXAMIN_BEGIN_DE']][['RESPOND_ID', 'TOUR_BEGIN_DE', 'EXAMIN_BEGIN_DE']].head(10)
```

```python
#동일 응답자 고성 속성 충돌 여부 확인 (성별, 연령대)

# 같은 응답자인데 성별이 여러 값으로 응답된 경우
sex_per_respondent = df.groupby('RESPOND_ID')['SEXDSTN_FLAG_CD'].nunique()
count_sex_conflict = (sex_per_respondent > 1).sum()
print(f'동일 응답자-성별 충돌: {count_sex_conflict}건')

# 같은 응답자인데 연령대가 2단계 이상 차이나는 경우
age_order_map = {'20대': 1, '30대': 2, '40대': 3, '50대': 4, '60대 이상': 5}
df['AGE_NUM'] = df['AGRDE_FLAG_NM'].map(age_order_map)
age_range_per_respondent = df.groupby('RESPOND_ID')['AGE_NUM'].agg(['min', 'max'])
count_age_conflict = ((age_range_per_respondent['max'] - age_range_per_respondent['min']) >= 2).sum()
print(f'동일 응답자-연령대 2단계+ 충돌: {count_age_conflict}건')
```

```python
#물리적 자원 충돌 확인 ( 여행기간 겹치는 경우)

# 날짜/기간 컬럼 준비
pd_map = {'1박 2일': 2, '2박 3일': 3, '3박 4일': 4, '4박 5일': 5, '5박 6일': 6, '6박 7일 이상': 7}
df['TRIP_DAYS'] = df['TOUR_PD_VALUE'].map(pd_map)
df['TOUR_BEGIN_DATE'] = pd.to_datetime(df['TOUR_BEGIN_DE'], format='%Y%m%d', errors='coerce')
df['TOUR_END_DATE'] = df['TOUR_BEGIN_DATE'] + pd.to_timedelta(df['TRIP_DAYS'].fillna(1) - 1, unit='D')

count_trip_overlap = 0
for respondent_id, trips in df[df['TOUR_BEGIN_DATE'].notna()].groupby('RESPOND_ID'):
    if len(trips) < 2:
        continue
    trips = trips.sort_values('TOUR_BEGIN_DATE')
    begin_dates = trips['TOUR_BEGIN_DATE'].values
    end_dates = trips['TOUR_END_DATE'].values
    for i in range(len(trips) - 1):
        if end_dates[i] > begin_dates[i + 1]:
            count_trip_overlap += 1

print(f'여행기간 물리적 겹침: {count_trip_overlap}건')
```

```python
#중복 키 확인 
# 동일 응답자 + 동일 여행시작일인데 조사 회차마다 응답이 다른 경우
trip_group_sizes = df.groupby(['RESPOND_ID', 'TOUR_BEGIN_DE']).size()
duplicate_trip_groups = trip_group_sizes[trip_group_sizes > 1]
count_duplicate_trip_response = duplicate_trip_groups.sum() - len(duplicate_trip_groups)
print(f'동일 여행 중복 응답(잉여분): {count_duplicate_trip_response}건')

# 그 중 여행 지역까지 다르게 기록된 경우
region_per_trip = df.groupby(['RESPOND_ID', 'TOUR_BEGIN_DE'])['TOUR_CTPRVN_NM'].nunique()
count_region_conflict = (region_per_trip > 1).sum()
print(f'동일여행-지역 충돌: {count_region_conflict}건')
```

```python
# 각 손상 유형별로 삭제 대상 RESPOND_ID 모으기 / 삭제 대상 겹치는지 확인 
time_violation_ids = df[df['TOUR_BEGIN_DE'] > df['EXAMIN_BEGIN_DE']]['RESPOND_ID'].unique()
sex_conflict_ids = sex_per_respondent[sex_per_respondent > 1].index
age_conflict_ids = age_range_per_respondent[(age_range_per_respondent['max'] - age_range_per_respondent['min']) >= 2].index

print('시간모순 응답자 수:', len(time_violation_ids))
print('성별충돌 응답자 수:', len(sex_conflict_ids))
print('연령충돌 응답자 수:', len(age_conflict_ids))

overlap = set(time_violation_ids) & set(sex_conflict_ids)
print('시간모순+성별충돌 겹치는 응답자 수:', len(overlap))

#시간모순+성별충돌 겹치지 않음 
```

## 손상 레코드 처리 전략

<div style="font-size: 12px; line-height: 1.3;">

### 시간 순서 모순 (`TOUR_BEGIN_DE` > `EXAMIN_BEGIN_DE`): 47건 (0.33%)
- 처리: 행 삭제
- 근거: 조사일보다 미래에 여행 갔다는 건 논리적으로 불가능함, 어느 날짜가 오기입인지 알 수 없어 조사 설계상 존재할 수 없는 조합을 남겨두는 것보다 제거가 안전함

### 동일응답자-연령대 2단계+ 충돌: 5명 (10건)
- 처리: 해당 응답자 행 전체 삭제
- 근거: 연령대는 이번 분석의 핵심 변수(성수기·연령대별 쏠림 분석)라 애매한 값을 남기면 판단 자체가 왜곡될 위험이 있음, 손실이 미미해(0.07%) 제거가 안전함

### 동일응답자-성별 충돌: 34명 (81건)
- 처리: 처리하지 않음(원본 유지)
- 근거: 반복 설문 구조상 응답자 개인의 불변 속성이 아니라 회차별 응답값으로 볼 수 있음, 보조 변수라 핵심 분석(지역·지출)에 영향이 적어 데이터 손실 대비 실익이 낮음

### 여행기간 물리적 겹침: 109건 (0.76%)
- 처리: 처리하지 않음(원본 유지)
- 근거: 겹침 판정에 사용한 여행기간 근사치("6박 7일 이상"→7일)의 오차에서 비롯된 착시일 가능성이 있어, 실제 데이터 오류로 단정하기 어려움

### 동일 여행 중복 응답 (동일 `RESPOND_ID`+`TOUR_BEGIN_DE` 여러 조사 회차): 87건 (0.61%)
- 처리: 최신 조사회차(`EXAMIN_BEGIN_DE` 최댓값)만 유지
- 근거: 완전 삭제 시 정보 손실 발생, 반복 조사에서 응답자가 기억을 정정했을 가능성이 있는 최신 응답을 채택하는 것이 유일하게 근거를 댈 수 있는 기준
- 부수효과: 동일여행-지역 충돌(5건)도 같은 그룹의 부분집합이라 최신 회차만 남기면 자동 해결됨

```python
print('처리 전 행 수:', len(df))

# 1. 시간 순서 모순 삭제
df = df[df['TOUR_BEGIN_DE'] <= df['EXAMIN_BEGIN_DE']]
print('시간모순 삭제 후:', len(df))

# 2. 동일 여행 중복 응답 -> 최신 회차만 유지
latest_idx = df.groupby(['RESPOND_ID', 'TOUR_BEGIN_DE'])['EXAMIN_BEGIN_DE'].idxmax()
df = df.loc[latest_idx]
print('중복응답(최신회차만 유지) 후:', len(df))

# 3. 연령대 2단계+ 충돌 응답자 전체 삭제
age_order_map = {'20대': 1, '30대': 2, '40대': 3, '50대': 4, '60대 이상': 5}
age_num = df['AGRDE_FLAG_NM'].map(age_order_map)
age_range_per_respondent = age_num.groupby(df['RESPOND_ID']).agg(['min', 'max'])
conflicted_respondents = age_range_per_respondent[
    (age_range_per_respondent['max'] - age_range_per_respondent['min']) >= 2
].index
df = df[~df['RESPOND_ID'].isin(conflicted_respondents)]
print('연령충돌 삭제 후:', len(df))

# 4. 성별 충돌 / 여행기간 겹침 -> 처리하지 않음 (원본 유지, 별도 코드 없음)

print()
print('최종 행 수:', len(df))
```

- 원본 데이터의 정합성을 재검증하는 용도로도 활용 가능 (예: 세부 지출 합산액과 응답자가 답한 총비용을 비교)
