### 3.이상치 처리

```python
# 주요 범주형 변수 값 확인
for col in [
    'TOUR_CTPRVN_NM_x',
    'TOUR_SIGNGU_NM_x',
    'TOUR_PD_VALUE_x',
    'TOUR_PURPS_NM'
]:
    print(f'\n[{col}]')
    print(df[col].value_counts())
```

```python
# 분석에 사용할 주요 컬럼의 실제 값 확인
cols = [
    'TOUR_BEGIN_DE',          # 여행시작일자
    'TOUR_CTPRVN_NM_x',       # 여행시도명
    'TOUR_SIGNGU_NM_x',       # 여행시군구명
    'TOUR_PD_VALUE_x',        # 여행기간
    'TOUR_PURPS_NM',          # 여행목적
    'TOUR_TOT_CT_VALUE'       # 여행총비용
]

for col in cols:
    print(f'\n[{col}]')
    print(df[col].value_counts(dropna=False).head(10))
```

이상치 확인 결과

- 데이터 범위를 벗어난 비정상적인 범주값은 발견되지 않아서 별도의 이상치 삭제는 수행하지 않음
- '모름'은 이상치가 아니라 미응답 값이므로 관련 분석에서만 제외

### 5.파생변수 생성

#### 5-1.여행 월 파생변수 생성

```python
# 여행 시작일에서 월 추출
df['TOUR_MONTH'] = df['TOUR_BEGIN_DE'].dt.month

# 확인
df['TOUR_MONTH'].value_counts().sort_index()
```

#### 5-2.숙박일수 파생변수 생성

```python
# 여행기간을 숙박일수로 변환
period_map = {
    '1박 2일': 1,
    '2박 3일': 2,
    '3박 4일': 3,
    '4박 5일': 4,
    '5박 6일': 5,
    '6박 7일 이상': 6
}

df['숙박일수'] = df['TOUR_PD_VALUE_x'].map(period_map)

# 확인
df[['TOUR_PD_VALUE_x', '숙박일수']].head(10)
```

```python
# 숙박일수로 변환되지 않은 원본 값 확인
df.loc[
    df['숙박일수'].isnull(),
    ['TOUR_PD_VALUE_x', '숙박일수']
].value_counts(dropna=False)
```

#### 5-3.고지출 여부 파생변수 생성

```python
# 고지출 여부 파생변수 생성
# 우선 모든 행을 결측값으로 설정
df['고지출여부'] = pd.Series(
    pd.NA,
    index=df.index,
    dtype='boolean'
)

# 총비용이 '모름'이 아닌 행
known_expense = df['TOUR_TOT_CT_VALUE'] != '모름'

# 총비용이 30만원 이상이면 True, 미만이면 False
df.loc[known_expense, '고지출여부'] = (
    df.loc[known_expense, 'TOUR_TOT_CT_VALUE']
    .isin(['30~40만원 미만', '40만원 이상'])
)
```

```python
# 여행 총비용이 확인된 데이터만 고지출 분석에 사용
expense_df = df[df['고지출여부'].notna()].copy()
```

```python
df[
    ['TOUR_TOT_CT_VALUE', '고지출여부']
].value_counts(dropna=False)
```

#### 5-4.세부 지출 점수 파생변수 생성

```python
# 점수로 변환할 세부 지출 항목
expense_cols = [
    'TOUR_LDGMNT_CT_VALUE',    # 숙박비
    'TOUR_FOOD_CT_VALUE',      # 음식비
    'TOUR_TRNSPORT_CT_VALUE',  # 교통비
    'TOUR_SHOPNG_CT_VALUE',    # 쇼핑비
    'TOUR_ACTVTY_CT_VALUE',    # 액티비티비
    'TOUR_ETC_CT_VALUE'        # 기타비용
]
```

```python
# 세부 지출 구간을 비교하기 위한 점수
expense_score = {
    '1만원 미만': 1,
    '1~3만원 미만': 2,
    '3~5만원 미만': 3,
    '5~7만원 미만': 4,
    '7~10만원 미만': 5,
    '10만원 이상': 6
}
```

```python
# 세부 지출 항목을 1~6점으로 변환
for col in expense_cols:
    df[col + '_SCORE'] = df[col].map(expense_score)
```

```python
# 숙박비를 예시로 원본 값과 점수 비교
df[[
    'TOUR_LDGMNT_CT_VALUE',
    'TOUR_LDGMNT_CT_VALUE_SCORE'
]].head(10)
```
